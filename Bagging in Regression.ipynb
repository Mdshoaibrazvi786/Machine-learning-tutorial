```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\mr.shoaib\Downloads\placement.csv")
```

```python
df.head(3)
```

```python
x = df.iloc[:,:-1]
y = df["package"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import BaggingRegressor
```

```python
bg = BaggingRegressor(estimator=LinearRegression(),n_estimators=40)
bg.fit(x_train, y_train)
```

```python
bg.score(x_train, y_train)*100
```

```python
bg.score(x_test, y_test)*100
```

```python
rf = RandomForestRegressor(n_estimators=50)
rf.fit(x_train,y_train)
```

```python
rf.score(x_train, y_train)*100
```

```python
rf.score(x_test, y_test)*100
```

```python
prd = {"bg":bg.predict(x_test),"rf":rf.predict(x_test)}
```
