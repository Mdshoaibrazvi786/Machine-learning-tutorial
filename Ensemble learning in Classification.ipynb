```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
```

```python
x,y = make_moons(n_samples=500, noise=0.2)
```

```python
df = {"x1":x[:,0],"x2":x[:,1],"y":y}
```

```python
df = pd.DataFrame(df)
```

```python
df
```

```python
sns.scatterplot(x="x1",y="x2",data=df,hue="y")
plt.show()
```

```python
x_a = df.iloc[::-1]
y_a = df["y"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_a,y_a,test_size=0.25,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.svm import SVC
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)
```

```python
dt.score(x_train, y_train)*100
```

```python
dt.score(x_test, y_test)*100
```

```python
sv = SVC()
sv.fit(x_train, y_train)
```

```python
sv.score(x_train, y_train)*100
```

```python
sv.score(x_test, y_test)*100
```

```python
gnb = GaussianNB()
gnb.fit(x_train, y_train)
```

```python
gnb.score(x_train, y_train)*100
```

```python
gnb.score(x_test, y_test)*100
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
li = [("dt1",DecisionTreeClassifier()),("sv1",SVC()),("gnb1",GaussianNB())]
```

```python
vc = VotingClassifier(li,weights=[5,4,7])
vc.fit(x_train, y_train)
```

```python
vc.score(x_train, y_train)*100
```

```python
vc.score(x_test, y_test)*100
```

```python
prd = {"dt":dt.predict(x_test),"svm":sv.predict(x_test),"gnb":gnb.predict(x_test),"vc":vc.predict(x_test)}
```
