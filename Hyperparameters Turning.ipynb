```python
import pandas as pd
```

```python
df =  pd.read_csv(r"C:\Users\mr.shoaib\Downloads\Salary.csv")
```

```python
df
```

```python
x =  df[['level']]
y =  df['Salary']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,random_state=42, test_size=0.2) 
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train, y_train)
```

```python
dt.score(x_train, y_train)*100, dt.score(x_test, y_test)*100
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
dt = {"criterion":['squared_error','friedman_mse','absolute_error','poisson'],
      "splitter":['best','random'],
      "max_depth":[i for i in range(2,20)]}
```

```python
df = GridSearchCV(DecisionTreeRegressor(),param_grid=dt)
```

```python
df.fit(x_train, y_train)
```

```python
df.best_params_
```

```python
df.best_score_
```

```python
rd = RandomizedSearchCV(DecisionTreeRegressor(),param_distributions=dt,n_iter=20)
rd.fit(x_train, y_train)
```

```python
rd.best_params_
```
