```python
import pandas as pd
```

```python
df=pd.DataFrame({"Size":["s","m","l","xl","s","m","l","s","s","l","xl","m"]})
df
```

```python
ord_data=[["s","m","l","xl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=ord_data)
oe.fit(df[['Size']])
```

```python
oe.transform(df[['Size']])
```

```python
df['Size_en'] = oe.transform(df[['Size']])
```
