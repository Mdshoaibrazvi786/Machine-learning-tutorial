```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\mr.shoaib\Downloads\flights.csv")
```

```python
df
```

```python
plt.scatter(df["year"],df["passengers"])
plt.show()
```

```python
# Convert to polynomial Feature With help of sk learn.
```

```python
x = df[["year"]]
y = df["passengers"]
```

```python
# Convert to polynomial Nature.
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
lr = PolynomialFeatures(degree=3) # hightest power ko batata hai.
lr.fit(x)
lr.transform(x)
```

```python
# Store To Variable.
```

```python
x = lr.transform(x)
```

```python
# Train Test split.
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Apply To Linear Regression.
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
lr.score(x_train,y_train)*100
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
prd = lr.predict(x)
```

```python
plt.scatter(df["year"],df["passengers"])
plt.plot(df["year"],prd,c='red')
plt.show()
```
