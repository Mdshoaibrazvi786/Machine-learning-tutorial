```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r"C:\Users\mr.shoaib\Downloads\flights.csv")
df
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="year",y="passengers",data=df)
plt.show()
```

```python
df.isnull().sum()
```

```python
x = df[["year"]] #ye input feature hai or 2 diamensional allow krta hai.
y = df["passengers"]
x
```

```python
y
```

```python
x_train,x_test,y_train,y_train = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()   #ye object hai.
lr.fit(x,y)
```

```python
lr.score(x,y)*100  #ye Accuracy hai or isko percentage me dekhane ke liye 100 se multiples kra hai.
```

```python
lr.predict([[112]])
```

```python
y_pre=lr.predict(x)
```

```python
plt.scatter(x,y,color="blue")
plt.plot(x,y_pre,color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
