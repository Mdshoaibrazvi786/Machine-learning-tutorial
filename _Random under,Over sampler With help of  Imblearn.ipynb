```python
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
df =  pd.read_excel(r"C:\Users\mr.shoaib\Documents\Social Network Ads.xlsx")
```

```python
df
```

```python
df['Purchased'].value_counts()
```

```python
x = df.iloc[:,:-1]
y = df['Purchased']
```

```python
#Apply to Random under sampler.
```

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
from imblearn.over_sampling import RandomOverSampler
```

```python
ro = RandomOverSampler()
ro_x, ro_y = ro.fit_resample(x,y)
```

```python
ro_x
```

```python
ro_y
```

```python
ro_y.value_counts()
```

```python
x_train1,x_test1,y_train1,y_test1=train_test_split(ro_x, ro_y,test_size=0.2,random_state=42)
```

```python
lr1 = LogisticRegression()
lr1.fit(x_train1, y_train1)
```

```python
lr1.score(x_test1, y_test1)*100
```

```python
lr1.predict([[19,190000]])
```

```python
ru = RandomUnderSampler()
ru_x, ru_y = ru.fit_resample(x,y)
```

```python
ru_y.value_counts()
```

```python
x_train,x_test,y_train,y_test=train_test_split(ru_x, ru_y,test_size=0.2,random_state=42)
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.predict([[67,22000]])
```
